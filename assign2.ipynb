```python
import random
```

```python
places = ['A', 'B', 'C']
dirt = {x: random.choice([0, 1]) for x in places}
print("State:",dirt)
pos = 0
for i in range(7):
    loc = places[pos]
    if dirt[loc]:
        print("At",loc, "Suck")
        dirt[loc] = 0
    else:
        print("At",loc, "Move to",places[(pos+1)%3])
        pos = (pos + 1) % 3
    print("State:",dirt)
    if random.random()<1:
        dirt[places[pos]]=1
print("Final:", dirt)
```

### 2.

```python
start = (8,1,3,0,2,6,4,7,5)
goal  = (6,8,3,1,0,2,4,7,5)

def h(state):
    return sum(state[i] != goal[i] and state[i] != 0 for i in range(9))

current = start
visited = {current}

while current != goal:
    empty = current.index(0)
    next_states = []

    for move in [-3, 3, -1, 1]:
        pos = empty + move

        if 0 <= pos < 9 and not (move == -1 and empty % 3 == 0) and not (move == 1 and empty % 3 == 2):
            state = list(current)
            state[empty], state[pos] = state[pos], state[empty]
            state = tuple(state)

            if state not in visited and h(state) <= h(current):
                next_states.append(state)

    if not next_states:
        print("No solution found")
        break

    current = min(next_states,key=h)
    visited.add(current)
    print(current)
```
