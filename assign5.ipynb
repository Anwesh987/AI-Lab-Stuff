```python
import heapq
import random

graph = {
    'A': [('B', 4), ('C', 2)],
    'B': [('D', 5), ('E', 10)],
    'C': [('D', 3), ('F', 8)],
    'D': [('G', 4), ('H', 6)],
    'E': [('G', 3), ('I', 7)],
    'F': [('H', 2), ('I', 5)],
    'G': [('I', 2)],
    'H': [('I', 3)],
    'I': []
}

def show_path(parent, start, goal):
    path = []
    node = goal

    while node != start:
        path.append(node)
        node = parent[node]

    path.append(start)
    path.reverse()
    return path


# 1. Greedy Best First Search

def greedy(start, goal):
    h = {node: random.randint(1, 10) for node in graph}
    h[goal] = 0

    pq = [(h[start], start)]
    visited = set()
    parent = {}

    while pq:
        cost, node = heapq.heappop(pq)

        if node in visited:
            continue

        visited.add(node)
        print("Visiting:", node)

        if node == goal:
            print("Path:", show_path(parent, start, goal))
            return

        for neighbour, weight in graph[node]:
            if neighbour not in visited:
                parent[neighbour] = node
                heapq.heappush(pq, (h[neighbour], neighbour))

    print("No path found")


# 2. A* Search

def astar(start, goal):
    h = {node: random.randint(1, 10) for node in graph}
    h[goal] = 0

    pq = [(h[start], 0, start)]
    dist = {start: 0}
    parent = {}

    while pq:
        f, g, node = heapq.heappop(pq)

        if g != dist[node]:
            continue

        print("Visiting:", node)

        if node == goal:
            print("Cost:", g)
            print("Path:", show_path(parent, start, goal))
            return

        for neighbour, weight in graph[node]:
            new_cost = g + weight

            if neighbour not in dist or new_cost < dist[neighbour]:
                dist[neighbour] = new_cost
                parent[neighbour] = node

                f = new_cost + h[neighbour]
                heapq.heappush(pq, (f, new_cost, neighbour))
    print("No path found")


print("GREEDY BEST FIRST SEARCH")
greedy('A', 'G')

print("\nA* SEARCH: A TO I")
astar('A', 'I')

print("\nA* SEARCH: B TO H")
astar('B', 'H')

print("\nA* SEARCH: C TO G")
astar('C', 'G')
```
