```python
def resolve(c1, c2):
    new_clauses = []

    for literal in c1:
        opposite = literal[1:] if literal.startswith('~') else '~' + literal

        if opposite in c2:
            new_clause = (c1 - {literal}) | (c2 - {opposite})

            # Ignore tautologies
            if not any(
                (x[1:] if x.startswith('~') else '~' + x) in new_clause
                for x in new_clause
            ):
                new_clauses.append(new_clause)

    return new_clauses


def resolution(clauses):
    while True:
        new_clauses = []

        for i in range(len(clauses)):
            for j in range(i + 1, len(clauses)):
                results = resolve(clauses[i], clauses[j])

                for clause in results:
                    if len(clause) == 0:
                        return True

                    if clause not in clauses and clause not in new_clauses:
                        new_clauses.append(clause)

        if not new_clauses:
            return False

        clauses.extend(new_clauses)


# Facts in CNF
clauses = [
    {'P'},
    {'~P', 'Q'},
    {'~Q', 'R'}
]

# Add the negation of the goal
clauses.append({'~R'})

if resolution(clauses):
    print("Goal is proved!")
else:
    print("Goal cannot be proved.")
```
