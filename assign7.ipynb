```python
from collections import deque

def water_jug():
    start = (0, 0)
    queue = deque([start])
    parent = {start: None}

    while queue:
        a, b = queue.popleft()

        if a == 2 or b == 2:
            path = []
            state = (a, b)

            while state is not None:
                path.append(state)
                state = parent[state]

            print("Solution:")
            for state in path[::-1]:
                print(state)
            return

        # All possible actions
        moves = [
            (3, b),                  # Fill jug A
            (a, 4),                  # Fill jug B
            (0, b),                  # Empty jug A
            (a, 0),                  # Empty jug B
            (a - min(a, 4-b), b + min(a, 4-b)),  # A to B
            (a + min(b, 3-a), b - min(b, 3-a))   # B to A
        ]

        for next_state in moves:
            if next_state not in parent:
                parent[next_state] = (a, b)
                queue.append(next_state)
water_jug()
```

```python
from collections import deque

def valid(m, c):
    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    # Check bank X
    if m > 0 and c > m:
        return False

    # Check bank Y
    if 3-m > 0 and 3-c > 3-m:
        return False

    return True


def missionaries_cannibals():
    start = (3, 3, 1)
    goal = (0, 0, 0)

    queue = deque([start])
    parent = {start: None}

    # Possible boat combinations
    moves = [(1, 0), (2, 0), (0, 1),
             (0, 2), (1, 1)]

    while queue:
        m, c, b = queue.popleft()

        if (m, c, b) == goal:
            path = []
            state = goal

            while state is not None:
                path.append(state)
                state = parent[state]

            print("Solution:")
            for state in path[::-1]:
                print(state)
            return

        for x, y in moves:
            if b == 1:
                new_state = (m-x, c-y, 0)
            else:
                new_state = (m+x, c+y, 1)

            nm, nc, nb = new_state

            if valid(nm, nc) and new_state not in parent:
                parent[new_state] = (m, c, b)
                queue.append(new_state)


missionaries_cannibals()
```
