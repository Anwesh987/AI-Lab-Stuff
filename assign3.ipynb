### 1.

```python
graph = {
    'A': [('B', 4, 8), ('C', 3, 5)],
    'B': [('D', 5, 10), ('E', 2, 9)],
    'C': [('D', 4, 8), ('E', 6, 6)],
    'D': [('F', 3, 9)],
    'E': [('F', 4, 7)],
    'F': []
}
paths = []
def find_paths(node, goal, visited, path, distance, scenic):
    if node == goal:
        paths.append((path, distance, scenic))
        return
    for next_node, d, s in graph[node]:
        if next_node not in visited:
            find_paths(next_node, goal, visited | {next_node},
                       path + [next_node], distance + d, scenic + s)
find_paths('A', 'F', {'A'}, ['A'], 0, 0)
shortest = min(paths, key=lambda x: x[1])
best_scenic = max(paths, key=lambda x: x[2])
print("Minimum distance:", shortest)
print("Maximum scenic quality:", best_scenic)
```

```python
from collections import deque

start = (8,1,3,0,6,2,4,7,5)
goal  = (6,8,3,1,0,2,4,7,5)
```

```python
def search(start, goal, dfs):
    frontier = [start] if dfs else deque([start])
    parent = {start: None}

    while frontier:
        current = frontier.pop() if dfs else frontier.popleft()

        if current == goal:
            break

        empty = current.index(0)

        for move in [-3, 3, -1, 1]:
            pos = empty + move

            if 0 <= pos < 9 and not (move == -1 and empty % 3 == 0) and not (move == 1 and empty % 3 == 2):
                next_state = list(current)
                next_state[empty], next_state[pos] = next_state[pos], next_state[empty]
                next_state = tuple(next_state)

                if next_state not in parent:
                    parent[next_state] = current
                    frontier.append(next_state)

    if goal not in parent:
        return None, len(parent)

    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    return path[::-1], len(parent)


dfs_path, dfs_states = search(start, goal, True)
bfs_path, bfs_states = search(start, goal, False)

print("DFS Steps:", len(dfs_path) - 1)
print("DFS States Explored:", dfs_states)

print("BFS Steps:", len(bfs_path) - 1)
print("BFS States Explored:", bfs_states)
```
