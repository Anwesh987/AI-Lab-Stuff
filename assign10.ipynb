```python
def apply_action(action):
    if action == "UNSTACK(B,A)":
        state.remove("ON(B,A)")
        state.remove("CLEAR(B)")
        state.remove("ARMEMPTY")
        state.add("HOLDING(B)")
        state.add("CLEAR(A)")

    elif action == "STACK(B,D)":
        state.remove("HOLDING(B)")
        state.remove("CLEAR(D)")
        state.add("ON(B,D)")
        state.add("CLEAR(B)")
        state.add("ARMEMPTY")

    elif action == "PICKUP(C)":
        state.remove("ONTABLE(C)")
        state.remove("CLEAR(C)")
        state.remove("ARMEMPTY")
        state.add("HOLDING(C)")

    elif action == "STACK(C,A)":
        state.remove("HOLDING(C)")
        state.remove("CLEAR(A)")
        state.add("ON(C,A)")
        state.add("CLEAR(C)")
        state.add("ARMEMPTY")

    print(action, "->", sorted(state))


# Initial state
state = {
    "ON(B,A)", "ONTABLE(A)", "ONTABLE(C)", "ONTABLE(D)",
    "CLEAR(B)", "CLEAR(C)", "CLEAR(D)", "ARMEMPTY"
}

# Goal stack
stack = [
    ("GOAL", "ON(C,A)"),
    ("GOAL", "ON(B,D)")
]

while stack:
    item = stack.pop()
    kind, value = item

    if kind == "ACTION":
        apply_action(value)

    elif value in state:
        continue

    elif value == "ON(B,D)":
        stack.append(("ACTION", "STACK(B,D)"))
        stack.append(("GOAL", "HOLDING(B)"))
        stack.append(("GOAL", "CLEAR(D)"))

    elif value == "HOLDING(B)":
        stack.append(("ACTION", "UNSTACK(B,A)"))

    elif value == "ON(C,A)":
        stack.append(("ACTION", "STACK(C,A)"))
        stack.append(("GOAL", "HOLDING(C)"))
        stack.append(("GOAL", "CLEAR(A)"))

    elif value == "HOLDING(C)":
        stack.append(("ACTION", "PICKUP(C)"))

print("\nGoal achieved:", "ON(B,D)" in state and "ON(C,A)" in state)
```
