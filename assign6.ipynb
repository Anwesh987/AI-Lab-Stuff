```python
import random

def display(board):
    for i in range(0, 9, 3):
        print(" | ".join(
            board[j] if board[j] != ' ' else str(j+1)
            for j in range(i, i+3)
        ))
        if i < 6:
            print("---------")


def winner(board):
    wins = [(0,1,2), (3,4,5), (6,7,8),
            (0,3,6), (1,4,7), (2,5,8),
            (0,4,8), (2,4,6)]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] and board[a] != ' ':
            return board[a]

    return None


def minimax(board, turn):
    result = winner(board)

    if result == 'X':
        return 10
    if result == 'O':
        return -10
    if ' ' not in board:
        return 0

    if turn == 'X':
        best = -float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                score = minimax(board, 'O')
                board[i] = ' '
                best = max(best, score)
        return best

    else:
        best = float('inf')

        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                score = minimax(board, 'X')
                board[i] = ' '
                best = min(best, score)

        return best


def ai_move(board, ai):
    best_score = -float('inf') if ai == 'X' else float('inf')
    move = -1

    for i in range(9):
        if board[i] == ' ':
            board[i] = ai
            score = minimax(board, 'O' if ai == 'X' else 'X')
            board[i] = ' '

            if ai == 'X' and score > best_score:
                best_score = score
                move = i

            elif ai == 'O' and score < best_score:
                best_score = score
                move = i

    return move


def play_game(board, turn, human):
    ai = 'O' if human == 'X' else 'X'

    while True:
        print()
        display(board)

        if winner(board):
            print(winner(board), "wins!")
            break

        if ' ' not in board:
            print("Draw!")
            break

        if turn == human:
            while True:
                try:
                    pos = int(input("Enter position (1-9): ")) - 1

                    if 0 <= pos < 9 and board[pos] == ' ':
                        board[pos] = human
                        break

                    print("Invalid position!")

                except ValueError:
                    print("Enter a number from 1 to 9.")

        else:
            pos = ai_move(board, ai)
            board[pos] = ai
            print("AI plays:", pos + 1)

        turn = 'O' if turn == 'X' else 'X'


# First game: Empty board

human = input("Choose X or O: ").upper()

while human not in ['X', 'O']:
    human = input("Choose X or O: ").upper()

first = random.choice(['X', 'O'])
print("First player:", first)

board = [' '] * 9
play_game(board, first, human)


# Second game: Random state with one move each

print("\nSECOND GAME")

board = [' '] * 9
first = random.choice(['X', 'O'])

positions = random.sample(range(9), 2)
board[positions[0]] = first
board[positions[1]] = 'O' if first == 'X' else 'X'

print("Random starting state:")
display(board)

print("Next player:", first)
play_game(board, first, human)
```
