### 1.

```python
import random
n = 4
board = [random.randint(0, 3) for _ in range(n)]
def conflicts(b):
    count = 0
    for i in range(n):
        for j in range(i+1, n):
            if b[i] == b[j] or abs(b[i]-b[j]) == abs(i-j):
                count += 1
    return count
print("Initial:", board, "Attacks:", conflicts(board))
while conflicts(board) > 0:
    best = board[:]
    for col in range(n):
        for row in range(n):
            temp = board[:]
            temp[col] = row
            if conflicts(temp) < conflicts(best):
                best = temp
    if best == board:
        break
    board = best
    print("Board:", board, "Attacks:", conflicts(board))
print("Final Board:", board)
print("Minimum Attacks:", conflicts(board))
```

### 2.

```python
import random, math
n = 16
board = [random.randint(0, 3) for _ in range(n)]
def conflicts(b):
    count = 0
    for i in range(n):
        for j in range(i+1, n):
            if b[i] == b[j] or abs(b[i]-b[j]) == abs(i-j):
                count += 1
    return count
t = 100
while t > 0.01 and conflicts(board) > 0:
    col = random.randint(0, 3)
    row = random.randint(0, 3)
    temp = board[:]
    temp[col] = row
    diff = conflicts(temp) - conflicts(board)
    if diff < 0 or random.random() < math.exp(-diff/t):
        board = temp
    t *= 0.95
    print("Board:", board, "Attacks:", conflicts(board))
print("Final Board:", board)
print("Minimum Attacks:", conflicts(board))
```
